Understanding the 'polygon annotations' and what they are actually representing

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

# let us select one annotation file
label_file = 'Plant-disease-1/train/labels/Powdery_Mildew_Squash_jpg.rf.83ab61fde6456847c2533e26059e909d.txt'

# and it's corresponding image
img_folder = 'Plant-disease-1/train/images'

base_name = os.path.splitext(os.path.basename(label_file))[0]

img_file = None
for ext in ['.jpg','.jpeg','.png','.JPG','.JPEG','.PNG']:
  candidate = os.path.join(img_folder, base_name + ext)
  if os.path.exists(candidate):
    img_file = candidate
    break

print('Image:',img_file)
print('Label:',label_file)

# Overlaying the Annotations
img = cv2.imread(img_file)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

height, width = img.shape[:2]

# Reading annotation
with open(label_file, 'r') as f:
  lines = f.readlines()

# Drawing every polygon
for line in lines:
  values = list(map(float, line.strip().split()))
  class_id = int(values[0])
  coords = values[1:]

  #converting normalized coordinates into pixel coordinates
  points = []
  for i in range(0, len(coords), 2):
    x = int(coords[i]*width)
    y = int(coords[i+1]*height)
    points.append([x,y])

  points = np.array(points, dtype = np.int32)

  # Drawing Polygon
  cv2.polylines(img, [points], isClosed = True, color=(255, 255, 0), thickness = 3)

  # Drawing points to see the actual annotation vertices
  for x,y in points:
    cv2.circle(img, (x,y), 5, (255, 255, 0), -1)

plt.figure(figsize=(12,8))
plt.imshow(img)
plt.axis('off')
plt.show()
```

Understanding whether all the annotation boxes have the same structure

```python
import os

label_folder = 'Plant-disease-1/train/labels'

total_files = 0
rect_like = 0
other_format = 0

for filename in os.listdir(label_folder):

  if not filename.endswith('.txt'):
    continue

  total_files += 1

  with open(os.path.join(label_folder, filename), 'r') as f:
    lines = f.readlines()

  for line in lines:
    values = list(map(float, line.strip().split()))

    if len(values) == 9:  #9 values = class + 4(x,y) points
      rect_like += 1
    else:
      other_format += 1

print('Total annotation files:', total_files)
print('9-value annotation rows:', rect_like)
print('Other format:', other_format)
```

Verifying annotations as rectangles

```python
from numpy._core.defchararray import endswith
import os
import math

label_folder = 'Plant-disease-1/train/labels'

total_polygons = 0
rectangle_polygons = 0
nonRect_polygons = 0

nonRect_exa = []

TOL = 1e-5

for filename in os.listdir(label_folder):
  if not filename.endswith('.txt'):
    continue

  filepath = os.path.join(label_folder, filename)

  with open(filepath, 'r') as f:
    lines = f.readlines()

  for line in lines:
    values = list(map(float, line.strip().split()))
    if len(values) != 9:
      continue

    total_polygons += 1

    coords = values[1:]
    points = np.array([
        (coords[0],coords[1]),
        (coords[2],coords[3]),
        (coords[4],coords[5]),
        (coords[6],coords[7])
    ])

    x_axis = points[:, 0]
    y_axis = points[:, 1]
    unique_x = len(np.unique(np.round(x_axis,5)))
    unique_y = len(np.unique(np.round(y_axis,5)))

    if unique_x == 2 and unique_y == 2:
      rectangle_polygons += 1
    else:
      nonRect_polygons += 1

      if len(nonRect_exa) < 5:
        nonRect_exa.append((filename, points.tolist()))

    print('Total Polygons:', total_polygons)
    print('Rectangle-like Polygons:', rectangle_polygons)
    print('Non-Rectangle like Polygons:', nonRect_polygons)

    if total_polygons > 0:
      print('Rectangle Percentage:',
            round(rectangle_polygons/total_polygons*100, 2), '%')

    print('\nExamples of Non-Rectangle Polygons:')
    for example in nonRect_exa:
      print(example)
```

What exactly are SAM and Non-SAM regions?

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Reading the image
img_bgr = cv2.imread(test_img)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# YOLO box
x1, y1, x2, y2 = map(int, box)

# SAM mask
mask = sam_mask.astype(bool)

# Confirming that SAM mask matches image size
if mask.shape[:2] != img_rgb.shape[:2]:
    mask = cv2.resize(
        mask.astype(np.uint8),
        (img_rgb.shape[1], img_rgb.shape[0]),
        interpolation=cv2.INTER_NEAREST
    ).astype(bool)

# YOLO Bounding Box Mask
box_mask = np.zeros(mask.shape, dtype=bool)
box_mask[y1:y2, x1:x2] = True

# Healthy part = SAM Pixels inside YOLO Bounding Box
healthy_mask = mask & box_mask

# Non-Healthy region = YOLO Box - SAM Mask
suspected_mask = box_mask & ~mask

print("YOLO box pixels:", np.count_nonzero(box_mask))
print("SAM/Healthy pixels:", np.count_nonzero(healthy_mask))
print("Non-Healthy/Suspected non-SAM pixels:", np.count_nonzero(suspected_mask))

print("Candidate non-SAM percentage:",
      round(
        np.count_nonzero(suspected_mask) /
        np.count_nonzero(box_mask) * 100,
        2
    ),"%"
)
```

- '**How different is every pixel in the YOLO box from the color distribution of the healthy SAM region?**'

```python
# Generating a damaged candidate map
import numpy as np
import cv2
import matplotlib.pyplot as plt

# HSV image
hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV).astype(np.float32)

# Healthy reference = SAM pixels inside YOLO box
healthy_pixels = hsv[healthy_mask]

# Calculate median HSV of healthy region
healthy_median = np.median(healthy_pixels, axis=0)

print("Healthy reference HSV median:")
print("Hue:", healthy_median[0])
print("Saturation:", healthy_median[1])
print("Value:", healthy_median[2])

# Extract all pixels inside YOLO box
box_pixels = hsv[box_mask]

# HSV distance from healthy reference
# Hue is circular, so calculate circular difference
hue_diff = np.abs(box_pixels[:, 0] - healthy_median[0])
hue_diff = np.minimum(hue_diff, 180 - hue_diff)

sat_diff = np.abs(box_pixels[:, 1] - healthy_median[1])
val_diff = np.abs(box_pixels[:, 2] - healthy_median[2])

# Combined distance
distance = np.sqrt(
    hue_diff**2 +
    sat_diff**2 +
    val_diff**2
)

print("\nHSV distance statistics:")
print("Min:", distance.min())
print("Max:", distance.max())
print("Mean:", distance.mean())
print("Median:", np.median(distance))

# Visualizing
distance_map = np.zeros(mask.shape, dtype=np.float32)
distance_map[box_mask] = distance

plt.figure(figsize=(7, 7))
plt.imshow(distance_map, cmap="jet")
plt.colorbar(label="HSV Distance from Healthy Reference")
plt.title("Candidate Damage Map")
plt.axis('off')
plt.show()
```

```python
# calculating affected areas at multiple thresholds

thresholds = [60, 80, 100, 120, 140]

# Only evaluate pixels inside the YOLO box
box_distances = distance_map[box_mask]

box_area = np.count_nonzero(box_mask)

print("YOLO box area:", box_area)
print()

for threshold in thresholds:

    affected_pixels = np.sum(box_distances >= threshold)

    percentage = affected_pixels / box_area * 100

    print(
        f"Threshold {threshold}: "
        f"{affected_pixels} pixels | "
        f"{percentage:.2f}% of YOLO box"
    )
```

- the high-distance pixels spatially correspond to the visible brown/discoloured region.

```python
# Leaf area reconstruction

# Healthy reference values from SAM
H_med = healthy_median[0]
S_med = healthy_median[1]
V_med = healthy_median[2]

# Calculating HSV differences for EVERY pixel inside YOLO box

roi_hsv = hsv[box_mask]

h_diff = np.abs(roi_hsv[:, 0] - H_med)
h_diff = np.minimum(h_diff, 180 - h_diff)

s_diff = np.abs(roi_hsv[:, 1] - S_med)
v_diff = np.abs(roi_hsv[:, 2] - V_med)

# Instead of one combined distance, looking at each channel separately

print("Healthy reference:")
print(f"H = {H_med}")
print(f"S = {S_med}")
print(f"V = {V_med}")

print("\nChannel differences:")
print(
    "Hue   Median:",
    np.median(h_diff),
    "| mean:",
    np.mean(h_diff)
)

print(
    "Sat   Median:",
    np.median(s_diff),
    "| mean:",
    np.mean(s_diff)
)

print(
    "Value median:",
    np.median(v_diff),
    "| mean:",
    np.mean(v_diff)
)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Create image-sized masks
h, w = img.shape[:2]

# YOLO bounding box
x1, y1, x2, y2 = map(int, box)

# Keep everything inside YOLO box
box_mask = np.zeros((h, w), dtype=np.uint8)
box_mask[y1:y2, x1:x2] = 1

# Make sure SAM mask is binary
sam_binary = (sam_mask > 0).astype(np.uint8)

# 2. Initialize GrabCut mask
grabcut_mask = np.full(
    (h, w),
    cv2.GC_BGD,
    dtype=np.uint8
)

# Outside YOLO box = definite background
grabcut_mask[box_mask == 0] = cv2.GC_BGD

# SAM-selected pixels = definite foreground
grabcut_mask[sam_binary == 1] = cv2.GC_FGD

# 3. Mark pixels around SAM as probable foreground
kernel = np.ones((15, 15), np.uint8)

expanded_sam = cv2.dilate(
    sam_binary,
    kernel,
    iterations=2
)

# Only allow probable foreground inside YOLO box
probable_fg = (
    (expanded_sam == 1) &
    (box_mask == 1) &
    (sam_binary == 0)
)

grabcut_mask[probable_fg] = cv2.GC_PR_FGD

# 4.Marking the remaining YOLO-box pixels as probable background rather than definite background
probable_bg = (
    (box_mask == 1) &
    (grabcut_mask == cv2.GC_BGD)
)

grabcut_mask[probable_bg] = cv2.GC_PR_BGD

# 5. Run GrabCut
bgd_model = np.zeros((1, 65), np.float64)
fgd_model = np.zeros((1, 65), np.float64)

cv2.grabCut(
    img,
    grabcut_mask,
    None,
    bgd_model,
    fgd_model,
    5,
    cv2.GC_INIT_WITH_MASK
)

# 6. Convert GrabCut result to binary leaf mask
leaf_mask = np.where(
    (grabcut_mask == cv2.GC_FGD) |
    (grabcut_mask == cv2.GC_PR_FGD),
    1,
    0
).astype(np.uint8)

# Restrict again to YOLO box
leaf_mask = leaf_mask * box_mask

#7. Visualize
plt.figure(figsize=(7, 7))
plt.imshow(leaf_mask, cmap="gray")
plt.title("GrabCut Whole-Leaf Candidate")
plt.axis("off")
plt.show()
```

```python
# Create overlay
overlay = img.copy()

# Color the GrabCut-selected leaf region
overlay[leaf_mask == 1] = (
    0.5 * overlay[leaf_mask == 1] +
    0.5 * np.array([0, 255, 0])
).astype(np.uint8)

plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.title("GrabCut Leaf Overlay")
plt.axis("off")
plt.show()
```
